```python
import os
import re
import kagglehub
import pandas as pd
from sklearn.model_selection import train_test_split

path = kagglehub.dataset_download("snap/amazon-fine-food-reviews")
df_full = pd.read_csv(os.path.join(path, "Reviews.csv"))
df = df_full.sample(n=50000, random_state=42)

df = df.dropna(subset=["Text", "Score"]).copy()
df = df.drop_duplicates(subset=["Text"]).copy()

def clean_text(text):
    text = str(text).lower()
    text = re.sub(r"<[^>]+>", " ", text)
    text = re.sub(r"[^a-z0-9\s']", " ", text)
    text = re.sub(r"\s+", " ", text)
    return text.strip()

df["Clean_Text"] = df["Text"].apply(clean_text)
df = df[df["Clean_Text"] != ""].copy()
df["Sentiment"] = (df["Score"] >= 4).astype(int)

train_df, test_df = train_test_split(
    df,
    test_size=0.20,
    random_state=42,
    stratify=df["Sentiment"]
)

train_df.to_csv("alpha_train.csv", index=False)
test_df.to_csv("alpha_test.csv", index=False)

print("Saved alpha_train.csv:", len(train_df), "| alpha_test.csv:", len(test_df))
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
import time

# Load the train and test data
train_text = train_df["Clean_Text"].fillna("").astype(str)
test_text = test_df["Clean_Text"].fillna("").astype(str)
train_labels = train_df["Sentiment"]
test_labels = test_df["Sentiment"]

# Vectorize the text data using TF-IDF
vectorizer = TfidfVectorizer(max_features=5000)
train_tfidf = vectorizer.fit_transform(train_text)
test_tfidf = vectorizer.transform(test_text)

# Train a Multinomial Naive Bayes classifier on the TF-IDF features
nb = MultinomialNB()
start_time = time.perf_counter()
nb.fit(train_tfidf, train_labels)
nb_train_time = time.perf_counter() - start_time
print(f"Naïve Bayes training time: {nb_train_time:.4f}s")
test_predictions = nb.predict(test_tfidf)

# Evaluate the model
print("Accuracy:", accuracy_score(test_labels, test_predictions))
print("\nClassification Report:\n", classification_report(test_labels, test_predictions))
```

```python
import matplotlib.pyplot as plt
from sklearn.metrics import ConfusionMatrixDisplay

fig, ax = plt.subplots(figsize=(6, 5))

ConfusionMatrixDisplay.from_estimator(
    nb,
    test_tfidf,
    test_labels,
    display_labels=["Negative (0)", "Positive (1)"],
    cmap="Blues",
    values_format="d",
    ax=ax,
)

ax.set_title("Naïve Bayes Confusion Matrix")
plt.tight_layout()
plt.show()
```

```python
# Save trained model
import os, joblib
os.makedirs("models", exist_ok=True)
joblib.dump(vectorizer, "models/tfidf_vectorizer.joblib")
joblib.dump(nb, "models/naive_bayes.joblib")
```
